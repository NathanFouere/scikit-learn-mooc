````python
import pandas as pd

ames_housing = pd.read_csv("https://raw.githubusercontent.com/bilals/scikit-learn-mooc/main/datasets/ames_housing_no_missing.csv")
target_name = "SalePrice"
data = ames_housing.drop(columns=target_name)
target = ames_housing[target_name]
````

````python
numerical_features = [
    "LotFrontage", "LotArea", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF", "LowQualFinSF",
    "GrLivArea", "BedroomAbvGr", "KitchenAbvGr", "TotRmsAbvGrd", "Fireplaces",
    "GarageCars", "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch",
    "3SsnPorch", "ScreenPorch", "PoolArea", "MiscVal",
]

data_numerical = data[numerical_features]
print(data_numerical)
````

We will compare the generalization performance of a decision tree and a linear regression. For this purpose, we will create two separate predictive models and evaluate them by 10-fold cross-validation.

Thus, use sklearn.linear_model.LinearRegression and sklearn.tree.DecisionTreeRegressor to create the models. Use the default parameters for the linear regression and set random_state=0 for the decision tree.

Be aware that a linear model requires to scale numerical features. Please use sklearn.preprocessing.StandardScaler so that your linear regression model behaves the same way as the quiz author intended ;)

````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_validate
````

````python
pipelineLinear = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LinearRegression()),
])

fit = pipelineLinear.fit(data_numerical, target)
scoreLinear = fit.score(data_numerical, target)
valLinear = cross_validate(pipelineLinear, data_numerical, target, cv=10)
print(scoreLinear, valLinear)
````

````python
pipelineDecisionTree = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    # ("scaler", StandardScaler()), Pas besoin de scaler: https://datascience.stackexchange.com/questions/97869/standardizing-giving-worse-results
    ("model", DecisionTreeRegressor(random_state=0)),
])

fit = pipelineDecisionTree.fit(data_numerical, target)
scoreDecisionTree = fit.score(data_numerical, target)
valDecisionTree = cross_validate(pipelineDecisionTree, data_numerical, target, cv=10)
print(scoreDecisionTree, valDecisionTree)
````

Question 1

By comparing the cross-validation test scores for both models fold-to-fold, count the number of
times the linear model has a better test score than the decision tree model.
Select the range which this number belongs to:

- a) [0, 3]: the linear model is substantially worse than the decision tree

````python
nb_linear_higher=0
nb_decision_tree_higher=0

for i in range(10):
  if valLinear['test_score'][i] > valDecisionTree['test_score'][i]:
    nb_linear_higher += 1
  else:
    nb_decision_tree_higher += 1

print("nblinearhigher:", nb_linear_higher, "nbdecisiongiher:", nb_decision_tree_higher)
````

Réponse: C


Instead of using the default parameters for the decision tree regressor, we will optimize the max_depth of the tree. Vary the max_depth from 1 level up to 15 levels. Use nested cross-validation to evaluate a grid-search (sklearn.model_selection.GridSearchCV). Set cv=10 for both the inner and outer cross-validations, then answer the questions below.

````python
from sklearn.model_selection import GridSearchCV

pipelineDecisionTree = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", DecisionTreeRegressor(random_state=0)),
])

params = [
    {
        "model__max_depth": range(1, 15)
    }
]

clf = GridSearchCV(param_grid=params, estimator=pipelineDecisionTree, cv=10)
outer = cross_validate(clf, data_numerical, target, cv=10)
fit = clf.fit(data_numerical, target)
print(fit.best_estimator_, fit.best_score_, outer)
````

Create a preprocessor by dealing separately with the numerical and categorical columns. For the sake of simplicity, we will assume the following:

    categorical columns can be selected if they have an object data type;
    use an OrdinalEncoder to encode the categorical columns;
    numerical columns should correspond to the numerical_features as defined above. This is a subset of the features that are not an object data type.

In addition, set the max_depth of the decision tree to 7 (fixed, no need to tune it with a grid-search).

Evaluate this model using cross_validate as in the previous questions.

````python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
object_columns=data.select_dtypes("object").columns.to_numpy()

numerical_columns=data.select_dtypes("number").columns.to_numpy()

pipeline_num = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
])

pipeline_cat = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(unknown_value=-1, handle_unknown="use_encoded_value")),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", pipeline_num, numerical_columns),
        ("cat", pipeline_cat, object_columns),
    ]
)

pipeline1 = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", DecisionTreeRegressor(max_depth=7, random_state=0))
])

pipeline2 = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", DecisionTreeRegressor(max_depth=7, random_state=1))
])

pipeline3 = Pipeline(steps=[
    ("preprocess", preprocessor),
    ("model", DecisionTreeRegressor(max_depth=7, random_state=2))
])



result1 = cross_validate(pipeline1, data, target, cv=10)
result2 = cross_validate(pipeline2, data, target, cv=10)
result3 = cross_validate(pipeline3, data, target, cv=10)
````
